1.Armstrong Number

```python
#Finding whether given input number is Armstrong Number or not
num=int(input("Enter any Number:"))
temp=num
digits=len(str(num))
total=0
while temp>0:
  order=temp%10
  total+=order**digits
  temp//=10
if total==num:
  print("Given number is Armstrong.")
else:
  print("Given number is not a Armstrong")
```

2.Finding GCD of two numbes

```python
#Finding the GCD of two numbers
a=int(input("Enter First number:"))
b=int(input("Enter Second number:"))
while b!=0:
  rem=a%b     #a>b
  a=b
  b=rem
print("GCD:",a)
```
