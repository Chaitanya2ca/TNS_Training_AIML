15.Write a program that counts the total number of vowels in a string entered by the user.

```python
text=str(input("Enter the text:"))
count=0                            #Intially we are keeping count to 0
for character in text.lower():     #We are converting the text into lower case and then we will find the vowels
  if character in "a,e,i,o,u":     #Using Membership operator to check whether it is vowel or not
    count +=1                      #Updating the count
print("Total Number of Vowels:",count)
```

17.Write a program that checks whether a given word is a palindrome.

Example:

```python
word=str(input("Enter the Word:"))
print(word)
rev=word[::-1]
if word==rev:
  print("Palindrome")
else:
  print("Not a Palindrome")
```

19.Write a program that converts every alternate character of a string into uppercase.

```python
text=input("Enter your text:")
result=''                           # Creating empty string named result
for i in range(len(text)):          # len(text) Finds the number of characters in text with in the range
  if i % 2 == 0:                    # Using modulus operator to find the position
    result += text[i].upper()       # if i is even we will convert that character into Uppercase
  else:
    result += text[i].lower()       # if i is odd we will convert that character into lowercase
print(result)
```
