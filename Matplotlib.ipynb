Exercise :

Use below dataset:

Students=["A","B","C","D","E"]

math=[80,65,90,70,85]

science=[75,70,95,60,88]

Create a graph comparing math and Science marks

Try to include:

Two lines, Different Markers, Different Colors, Legend, Title, x/y grids, Grid

```python
#importing libraries
import matplotlib.pyplot as plt

#Dataset

students = ["A","B","C","D","E"]
math = [80,65,90,70,85]
science = [75,70,95,60,88]

#Creating Graphs
plt.plot(students,math, color = 'red',marker = 's',label = "Math")
plt.plot(students,science,color = 'green',marker = '*', label = "Science")

plt.title("Comparision between Maths and Science")
plt.legend()
plt.xlabel("Students")
plt.ylabel("Marks")
plt.grid(True)
plt.show()
```

1.Write a Python program to create a bar chart showing the number of books read by 5 students:

Students = ["A", "B", "C", "D", "E"]

Books = [4, 7, 3, 6, 5]

```python
#Importing library
import matplotlib.pyplot as plt

#Dataset
Students = ["A","B","C","D","E"]
Books = [4,7,3,6,5]

#Creating the graph
plt.bar(Students,Books)
plt.title("Bar Graph")
plt.xlabel("Students")
plt.ylabel("Books")
plt.show()
```

2.Write a Python program to create a pie chart showing how a student spends 10 hours studying:

Python = 4 hours

Math = 2 hours

Science = 2 hours

English = 1 hour

Other = 1 hour

```python
#Importing library
import matplotlib.pyplot as plt

#Dataset
subjects = ["Python","Math","Science","English","Other"]
hours = [4,2,2,1,1]

#Creating Pie chart
plt.pie(hours,labels=subjects,autopct='%1.1f%%')
plt.title("Pie Chart")
plt.show()
```

3.Write a Python program to create a scatter plot using the following data:

Study Hours = [1, 2, 3, 4, 5, 6]

Marks = [45, 50, 60, 65, 75, 85]

Add a title and labels.

```python
#Importing Libraries
import matplotlib.pyplot as plt

#Dataset
Study_Hours = [1,2,3,4,5,6]
Marks = [45,50,60,65,75,85]

#Creating Scatter plot
plt.scatter(Study_Hours,Marks)
plt.title("Scatter Plot")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.show()
```

5.Write a Python program to plot the performance of two students over 5 tests:

Tests = [1, 2, 3, 4, 5]

Student 1 = [60, 65, 70, 75, 80]

Student 2 = [55, 68, 72, 78, 85]

Display both lines on the same graph.

```python
#Importing Libraries
import matplotlib.pyplot as plt

#Dataset
Test = [1,2,3,4,5]
Student_1 = [60,65,70,75,80]
Student_2 = [55,68,72,78,85]

#Creating Line Chart
plt.plot(Test,Student_1, color = 'blue', marker = 'o', label = "Student_1")
plt.plot(Test,Student_2, color = 'red', marker = '*', label = "Student_2")
plt.title("Marks Comparision")
plt.xlabel("Test")
plt.ylabel("Markss")
plt.legend()
plt.show()
```
